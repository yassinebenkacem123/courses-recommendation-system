Avant d’appliquer un modèle de prédiction pour estimer les niveaux manquants, une étape heuristique basée sur la recherche de mots-clés a été mise en place. Cette étape consiste à analyser les champs textuels title, headline et description afin d’identifier la présence de mots-clés explicitement liés à un niveau pédagogique donné (par exemple : “beginner”, “intermediate”, “advanced”). Lorsqu’un mot-clé déterminant est détecté, le niveau correspondant est attribué avec une forte certitude, cette approche étant considérée comme fiable à 100 % dans ces cas précis. Les cours pour lesquels aucun mot-clé explicite n’est identifié seront ensuite traités par un modèle de prédiction automatique.

```python
keywords_level = {
    'Beginner': [
        'beginner', 'beginners', 'begginer', 'begginers','intro', 'introduction', 'introductions','basic', 'basics', 'fundamental', 'fundamentals',
        'start', 'starting', 'starter', 'get started','foundation', 'foundations', 'essential', 'essentials','101', 'level 1', 'part 1', 'chapter 1', 'module 1',
        'crash course', 'quick start', 'first steps', 'newbie', 'novice' ],

    'Intermediate': [
        'intermediate', 'intermediat', 'intermediet','mid-level', 'medium', 'beyond basics', 'advanced beginner', 'next level', 'step up',
        'level 2', 'part 2', 'chapter 2', 'module 2','deep dive', 'in-depth', 'comprehensive','professional', 'practitioner', 'intermediaire'
    ],

    'Advanced': [
        'advanced', 'advance', 'advancd', 'expert',  # "expert" ajouté ici
        'master', 'mastery', 'masters',
        'pro', 'professional', 'professional level',
        'guru', 'ninja', 'black belt', 'elite',
        'deep', 'deepest', 'ultimate', 'complete',
        'level 3', 'part 3', 'chapter 3', 'module 3',
        'specialist', 'specialization', 'certification'
    ],

    'All Levels': [
        'all levels', 'all-levels', 'any level',
        'everyone', 'complete guide', 'full course',
        'zero to hero', 'from beginner to advanced',
        'comprehensive', 'complete', 'all levels welcome',
        'beginners to advanced'
    ]
}

def remplir_level_direct(df):
    df_filled = df.copy()

    # Compter les niveaux remplis
    niveaux_trouves = {'Beginner': 0, 'Intermediate': 0, 'Advanced': 0, 'All Levels': 0}

    for idx, row in df_filled.iterrows():
        # Si le niveau est déjà rempli, passer au suivant
        if pd.notna(row['level']):
            continue

        titre = str(row['title']).lower()
        headline = str(row.get('headline', '')).lower()
        description = str(row.get('description', '')).lower()

        # Recherche du keyword dans titre, headline et description
        niveau_trouve = None

        # Priorité de recherche : titre > headline > description
        for text in [titre, headline, description]:
            if niveau_trouve:
                break
            for level, keywords in keywords_level.items():
                if any(keyword.lower() in text for keyword in keywords):
                    niveau_trouve = level
                    niveaux_trouves[level] += 1
                    break

        # Remplir la colonne level
        if niveau_trouve:
            df_filled.at[idx, 'level'] = niveau_trouve

    # Vérification détaillée
    total_rows = len(df_filled)
    filled_count = df_filled['level'].notna().sum()
    initial_na = df['level'].isna().sum()

    return df_filled

# Appliquer la fonction
data_level_manquants = remplir_level_direct(data_level_manquants)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

""" K-means"""
def optimized_clustering_pipeline(df):
    """
    Pipeline optimisé pour vos données spécifiques
    """
    print("="*80)
    print("CLUSTERING OPTIMISÉ - SANS REDONDANCE")
    print("="*80)
    
    # 1. Vérifier la structure
    print("\n[INFO] Colonnes disponibles:")
    print(df.columns.tolist())
    
    # 2. Features numériques
    print("\n[1/4] Extraction features numériques...")
    numerical_features = [
        'price', 'num_subscribers', 'avg_rating', 
        'num_reviews', 'num_lectures', 'content_length_min'
    ]
    X_numerical = df[numerical_features].copy()
    scaler = StandardScaler()
    X_num_scaled = scaler.fit_transform(X_numerical)
    print(f"✓ Shape: {X_num_scaled.shape}")
    
    # 3. Features catégorielles (INCLUENT déjà category, subcategory, topic)
    print("\n[2/4] Extraction features catégorielles...")
    categorical_features = ['category', 'subcategory', 'topic', 'level']
    
    X_cat_list = []
    encoders = {}
    for feature in categorical_features:
        le = LabelEncoder()
        encoded = le.fit_transform(df[feature])
        X_cat_list.append(encoded.reshape(-1, 1))
        encoders[feature] = le
        print(f"✓ {feature}: {len(le.classes_)} catégories")
    
    X_categorical = np.hstack(X_cat_list)
    scaler_cat = StandardScaler()
    X_cat_scaled = scaler_cat.fit_transform(X_categorical)
    print(f"✓ Shape: {X_cat_scaled.shape}")
    
    # 4. Features textuelles (UNIQUEMENT title + headline)
    print("\n[3/4] Extraction features textuelles (title + headline)...")
    df['text_content'] = (
        df['title'].fillna('') + ' ' + 
        df['headline'].fillna('')
    ).str.lower().str.strip()
    
    tfidf = TfidfVectorizer(
        max_features=150,
        stop_words='english',
        ngram_range=(1, 2),
        min_df=3,
        max_df=0.85
    )
    X_text = tfidf.fit_transform(df['text_content'])
    X_text_dense = X_text.toarray()
    print(f"✓ Shape: {X_text_dense.shape}")
    print(f"✓ Top termes: {tfidf.get_feature_names_out()[:10]}")
    
    # 5. Combinaison avec pondérations optimisées
    print("\n[4/4] Combinaison des features...")
    # Nouvelle pondération sans redondance
    X_combined = np.hstack([
        X_num_scaled * 0.4,    # 40% numérique
        X_cat_scaled * 0.3,    # 30% catégoriel
        X_text_dense * 0.3     # 30% textuel (sans redondance)
    ])
    print(f"✓ Shape finale: {X_combined.shape}")
    
    # 6. PCA
    print("\n[5/6] Réduction PCA...")
    pca = PCA(n_components=0.95)  # 95% variance
    X_pca = pca.fit_transform(X_combined)
    print(f"✓ Dimensions: {X_combined.shape[1]} → {X_pca.shape[1]}")
    print(f"✓ Variance expliquée: {pca.explained_variance_ratio_.sum():.4f}")
    
    # 7. Test multiple k
    print("\n[6/6] Test de différents k...")
    best_k = None
    best_sil = -1
    
    for k in range(3, 11):
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        labels = kmeans.fit_predict(X_pca)
        sil = silhouette_score(X_pca, labels)
        db = davies_bouldin_score(X_pca, labels)
        
        print(f"k={k}: Silhouette={sil:.4f}, Davies-Bouldin={db:.4f}")
        
        if sil > best_sil:
            best_sil = sil
            best_k = k
    
    print(f"\n🏆 Meilleur k: {best_k} (Silhouette: {best_sil:.4f})")
    
    # 8. Clustering final avec meilleur k
    kmeans_final = KMeans(n_clusters=best_k, random_state=42, n_init=10)
    labels_final = kmeans_final.fit_predict(X_pca)
    
    # Évaluation finale
    print("\n" + "="*80)
    print("RÉSULTATS FINAUX")
    print("="*80)
    sil_final = silhouette_score(X_pca, labels_final)
    db_final = davies_bouldin_score(X_pca, labels_final)
    ch_final = calinski_harabasz_score(X_pca, labels_final)
    
    print(f"Silhouette Score: {sil_final:.4f}")
    print(f"Davies-Bouldin Index: {db_final:.4f}")
    print(f"Calinski-Harabasz Score: {ch_final:.2f}")
    
    # Distribution
    unique, counts = np.unique(labels_final, return_counts=True)
    print(f"\nDistribution des clusters:")
    for cluster, count in zip(unique, counts):
        print(f"  Cluster {cluster}: {count} cours ({count/len(labels_final)*100:.2f}%)")
    
    return {
        'X_pca': X_pca,
        'pca': pca,
        'kmeans': kmeans_final,
        'labels': labels_final,
        'best_k': best_k,
        'metrics': {
            'silhouette': sil_final,
            'davies_bouldin': db_final,
            'calinski_harabasz': ch_final
        }
    }

# Utilisation
results = optimized_clustering_pipeline(data_finale)
```

```python
from difflib import get_close_matches

def find_course_index(df, title_query):
    titles = df['title'].astype(str)
    
    # Exact / contains match
    mask = titles.str.contains(title_query, case=False, na=False)
    if mask.sum() == 1:
        return df[mask].index[0]
    
    # Plusieurs résultats → meilleur match
    if mask.sum() > 1:
        return df[mask].index[0]
    
    # Fallback fuzzy matching
    matches = get_close_matches(title_query, titles, n=1, cutoff=0.6)
    if matches:
        return df[df['title'] == matches[0]].index[0]
    
    raise ValueError(f"Cours introuvable : '{title_query}'")
```

```python
import numpy as np

"""K-means """
def recommend_by_cluster_general(
    df,
    labels,
    course_title,
    n_recommendations=10,
    verbose=True
):
    """
    Recommandation générique basée sur le cluster + similarité métier
    """
    
    # 1. Trouver le cours source
    course_idx = find_course_index(df, course_title)
    course = df.loc[course_idx]
    cluster_id = labels[course_idx]
    
    # 2. Filtrer le cluster
    candidates = df[(labels == cluster_id) & (df.index != course_idx)].copy()
    
    # 3. Similarity score (générique)
    candidates['similarity_score'] = (
        (candidates['subcategory'] == course['subcategory']).astype(int) * 30 +
        (candidates['topic'] == course['topic']).astype(int) * 20 +
        (candidates['level'] == course['level']).astype(int) * 15 +
        (candidates['language'] == course['language']).astype(int) * 10 +
        (candidates['avg_rating'] / 5) * 20 +
        (np.log1p(candidates['num_subscribers']) /
         np.log1p(candidates['num_subscribers'].max())) * 15
    )
    
    # Bonus prix
    candidates['similarity_score'] += (
        (np.abs(candidates['price'] - course['price']) < 20).astype(int) * 10
    )
    
    # 4. Top-K
    recommendations = candidates.nlargest(n_recommendations, 'similarity_score')
    
    # 5. Affichage optionnel
    if verbose:
        print("="*80)
        print("📚 COURS SOURCE")
        print("="*80)
        print(f"{course['title']} | Cluster {cluster_id}")
        print(f"{course['category']} > {course['subcategory']} > {course['topic']}")
        
        print("\n🎯 RECOMMANDATIONS\n" + "="*80)
        for i, (_, c) in enumerate(recommendations.iterrows(), 1):
            print(f"{i}. {c['title']}  ⭐ {c['avg_rating']:.2f}  👥 {int(c['num_subscribers']):,}")
            print(f"   🎯 Score: {c['similarity_score']:.1f}")
            print()
    
    return recommendations
```

```python
# ==================== IMPORTS DEEP LEARNING ====================
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Deep Learning
import tensorflow as tf
from tensorflow.keras import layers, models, Model
from tensorflow.keras.callbacks import EarlyStopping

# ML
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics.pairwise import cosine_similarity

# ==================== 1. AUTOENCODEUR SIMPLE ====================

def build_autoencoder(input_dim, latent_dim=32):
    """Construit un autoencodeur simple"""
    # Encoder
    inputs = layers.Input(shape=(input_dim,))
    x = layers.Dense(128, activation='relu')(inputs)
    x = layers.Dense(64, activation='relu')(x)

    # Latent space
    latent = layers.Dense(latent_dim, activation='relu', name='latent')(x)

    # Decoder
    x = layers.Dense(64, activation='relu')(latent)
    x = layers.Dense(128, activation='relu')(x)
    outputs = layers.Dense(input_dim, activation='linear')(x)

    # Modèle complet
    autoencoder = Model(inputs, outputs)
    autoencoder.compile(optimizer='adam', loss='mse')

    # Modèle pour extraire embeddings
    encoder = Model(inputs, latent)

    return autoencoder, encoder

# ==================== 2. PRÉPARATION DES FEATURES ====================

def prepare_features_dl(df):
    """Prépare les features pour le deep learning"""
    print("🔧 Préparation des features...")

    # 1. Texte (TF-IDF + SVD)
    df['text_combined'] = (
        df['title'].fillna('') + ' ' +
        df['headline'].fillna('')
    ).str.lower().str.strip()

    tfidf = TfidfVectorizer(
        max_features=500,
        stop_words='english',
        ngram_range=(1, 2)
    )
    X_text = tfidf.fit_transform(df['text_combined'])

    # Réduction dimensionnelle
    svd = TruncatedSVD(n_components=50, random_state=42)
    X_text_svd = svd.fit_transform(X_text)

    # 2. Features numériques
    num_features = []
    for col in ['num_subscribers', 'avg_rating', 'num_reviews',
                'num_lectures', 'content_length_min', 'price']:
        if col in df.columns:
            num_features.append(col)

    scaler_num = MinMaxScaler()
    X_num = scaler_num.fit_transform(df[num_features])

    # 3. Features catégorielles
    cat_features = []
    for col in ['category', 'subcategory', 'topic']:
        if col in df.columns:
            le = LabelEncoder()
            df[col+'_encoded'] = le.fit_transform(df[col].fillna('unknown'))
            cat_features.append(col+'_encoded')

    if cat_features:
        scaler_cat = MinMaxScaler()
        X_cat = scaler_cat.fit_transform(df[cat_features])
    else:
        X_cat = np.array([])

    # 4. Combinaison
    if X_cat.size > 0:
        X_combined = np.hstack([X_text_svd, X_num, X_cat])
    else:
        X_combined = np.hstack([X_text_svd, X_num])

    print(f"✓ Features shape: {X_combined.shape}")

    return {
        'X': X_combined,
        'df': df.copy(),
        'tfidf': tfidf,
        'svd': svd,
        'scaler_num': scaler_num
    }

# ==================== 3. ENTRÂINEMENT AUTOENCODEUR ====================

def train_autoencoder(X, latent_dim=32, epochs=30):
    """Entraîne l'autoencodeur"""
    print("\n🤖 Entraînement de l'autoencodeur...")

    # Construire modèle
    autoencoder, encoder = build_autoencoder(
        input_dim=X.shape[1],
        latent_dim=latent_dim
    )

    # Callbacks
    callbacks = [EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)]

    # Entraînement
    history = autoencoder.fit(
        X, X,
        epochs=epochs,
        batch_size=64,
        validation_split=0.1,
        callbacks=callbacks,
        verbose=1
    )

    # Générer embeddings
    embeddings = encoder.predict(X, verbose=0)

    print(f"✓ Autoencodeur entraîné")
    print(f"✓ Embeddings shape: {embeddings.shape}")

    return {
        'autoencoder': autoencoder,
        'encoder': encoder,
        'embeddings': embeddings,
        'history': history.history
    }

# ==================== 4. RECOMMANDATION DEEP LEARNING ====================

class DLRecommender:
    """Système de recommandation Deep Learning"""

    def __init__(self, embeddings, df):
        self.embeddings = embeddings
        self.df = df.reset_index(drop=True)
        self.compute_similarity()

    def compute_similarity(self):
        """Calcule la matrice de similarité"""
        print("📊 Calcul des similarités...")
        self.similarity_matrix = cosine_similarity(self.embeddings)
        print(f"✓ Matrice de similarité: {self.similarity_matrix.shape}")

    def recommend(self, course_idx, top_n=10):
        """Recommandation par index"""
        if course_idx >= len(self.df):
            print(f"❌ Index {course_idx} hors limites")
            return []

        # Similarités
        sim_scores = self.similarity_matrix[course_idx].copy()
        sim_scores[course_idx] = -1

        # Top N
        top_indices = np.argsort(sim_scores)[::-1][:top_n]

        return top_indices

    def recommend_by_title(self, query_title, top_n=10):
        """Recommandation par titre"""
        # Chercher le titre
        query_lower = query_title.lower().strip()

        # Exact match d'abord
        exact_matches = self.df[self.df['title'].str.lower() == query_lower]

        if not exact_matches.empty:
            matches = exact_matches
        else:
            # Recherche partielle
            matches = self.df[
                self.df['title'].str.lower().str.contains(query_lower, na=False)
            ]

        if matches.empty:
            print(f"❌ Aucun cours trouvé avec '{query_title}'")
            return None

        # Prendre le premier
        course_idx = matches.index[0]
        course_title = self.df.loc[course_idx, 'title']

        print(f"\n🎯 COURS: {course_title}")
        print("="*60)

        # Obtenir recommandations
        recommended_indices = self.recommend(course_idx, top_n)

        # Afficher
        print(f"\n📚 TOP {top_n} RECOMMANDATIONS (Deep Learning):")
        print("-" * 80)

        for i, idx in enumerate(recommended_indices, 1):
            course = self.df.loc[idx]
            similarity = self.similarity_matrix[course_idx, idx]

            print(f"\n{i}. [{similarity:.3f}] {course['title']}")

            # Infos supplémentaires
            info = []
            if 'instructor_name' in course and pd.notna(course['instructor_name']):
                info.append(f"👤 {course['instructor_name']}")
            if 'avg_rating' in course and pd.notna(course['avg_rating']):
                info.append(f"⭐ {course['avg_rating']:.1f}/5.0")
            if 'num_subscribers' in course and pd.notna(course['num_subscribers']):
                info.append(f"👥 {int(course['num_subscribers']):,}")
            if 'category' in course and pd.notna(course['category']):
                info.append(f"🏷️ {course['category']}")

            if info:
                print(f"   {' | '.join(info)}")

        return recommended_indices

# ==================== 5. PIPELINE COMPLET SIMPLE ====================

"""Deep learning."""
def dl_recommendation_pipeline(df, query_title="Python", top_n=8):
    """
    Pipeline complet Deep Learning pour recommandations
    """
    print("="*60)
    print("🧠 SYSTÈME DE RECOMMANDATION DEEP LEARNING")
    print("="*60)

    # 1. Préparer features
    features = prepare_features_dl(df)
    X = features['X']

    # 2. Entraîner autoencodeur
    dl_results = train_autoencoder(X, latent_dim=32, epochs=30)
    embeddings = dl_results['embeddings']

    # 3. Initialiser recommender
    recommender = DLRecommender(embeddings, features['df'])

    # 4. Faire une recommandation
    print(f"\n🔍 Test avec: '{query_title}'")
    recommendations = recommender.recommend_by_title(query_title, top_n=top_n)

    return {
        'recommender': recommender,
        'embeddings': embeddings,
        'autoencoder': dl_results['autoencoder'],
        'encoder': dl_results['encoder']
    }

# ==================== 6. UTILISATION ====================

# Exécuter le pipeline
print("🚀 Initialisation du système Deep Learning...")
results = dl_recommendation_pipeline(
    df=data_clean,  # Votre DataFrame
    query_title="Python for Beginners",
    top_n=10
)

# Interface interactive
print("\n" + "="*60)
print("💬 INTERFACE INTERACTIVE")
print("="*60)

recommender = results['recommender']

while True:
    print("\nOptions:")
    print("1. Recommander par titre")
    print("2. Quitter")

    choice = input("\nVotre choix (1-2): ").strip()

    if choice == '1':
        title = input("Titre du cours: ").strip()
        if title:
            recommender.recommend_by_title(title, top_n=8)

    elif choice == '2':
        print("👋 Au revoir!")
        break
```
